### What is RMSProp?

### RMSProp (Root Mean Square Propagation)

### In AdaGrad  the learning rate becoming too small. RMSProp solves this by forgetting old gradients using an exponentially weighted moving average (EWMA). These preventing it from becoming too small.¶

### Instead of storing all past gradients, RMSProp keeps a moving average of recent squared gradients.

### RMSProp uses an exponentially weighted moving average of squared gradients to prevent AdaGrad’s learning rate from becoming too small.
