## Nesterov Accelerated Gradient (NAG) is an improved version of SGD with Momentum.
### NAG = Momentum + Lookahead Gradient Correction

#### NAG has two steps, but only one final parameter update.

## 1. Intuition (key idea)
 
 #### Momentum:
  - Take a step in the previous direction and then compute the gradient.

 ### NAG:
  - First look ahead in the direction of momentum,
then compute the gradient at that future position.
