## AdaGrad ?

# It automatically adjusts the learning rate for each parameter based on how often that parameter has been updated.

#### Core Idea (Simple Words)

**Parameters that receive large gradients often should slow down,
parameters that receive small or rare gradients should speed up.**

## Mathematical Intuition




### Note:
**Sparse data doesn’t create an elongated bowl, but it causes uneven learning dynamics that behave like one — and AdaGrad fixes this by adapting learning rates per parameter.**

## When to Use AdaGrad

#### Best Scenarios to Use AdaGrad

   **1️. Sparse Data**

   **2️. Different Feature Frequencies**

   **3️. Shallow / Linear Models**

   **4️. Fast Initial Convergence Needed**

   **5️. When You Don’t Want to Tune Learning Rate**


### When NOT to Use AdaGrad

#### Long Training / Deep Neural Networks
 - Learning rate keeps shrinking

 - Eventually becomes too small

 - Training may stall

***Use Adam or RMSProp instead***
