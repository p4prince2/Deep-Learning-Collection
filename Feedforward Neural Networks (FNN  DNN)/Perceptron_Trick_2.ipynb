*Before diving into the Perceptron Trick, it’s essential to fully understand the geometry of a line, especially since the perceptron tries to find a line (or hyperplane) that separates data points.*

## Let's Begin: What is a Line?
In 2D, a line is a set of all points (𝑥,𝑦) that satisfy a linear equation. 
 - The most general form is:
    -   <span style="color:red;">**Ax +By + c  = 0**</span>
  
Where:
- 𝐴,𝐵 are coefficients that define the direction,
- 𝐶 adjusts the position (intercept).    

# Visual Intuition

A line <span style="color:black;">**Ax +By + c  = 0**</span> divides the 2D space into two regions:
- <span style="color:green;">**Ax +By + c  > 0**</span>  Positive Side
- <span style="color:red;">**Ax +By + c  < 0**</span> Negative side

Changing the x (coefficient) affects the x-location, but y stays the same at the intercept.

## What Happens When You Change 𝐶 in the Line Equation:



**Here’s the graph illustrating how changing 𝐶 in the equation Ax +By + c = 0 affects the line's position:**

 - Green line : <span style="color:green;">**C = -4**</span> line shifts up into the positive region.
 - Black line :  **𝐶 = 0**  line passes through the origin.
 - Red line   : <span style="color:red;">**C = +4**</span> line shifts down into the negative region.

Increasing 𝐶 pushes the line toward the negative region, and decreasing it moves the line toward the positive region.


## What Happens When You Change A in the Line Equation:

| Action                                        | Effect on Output                                 |
| --------------------------------------------- | -------------------------------------------------|
| Increase weight `A` (in `Ax + By + C`)        | Line tilts clockwise (trun  positive(right))     |
| Decrease weight `A`                           | Line tilts counterclockwise (turn Negative(left))|

## What Happens When You Change B in the Line Equation

| Change in $B$ | Slope $-A/B$                   | Line Behavior                  |
| ------------- | ------------------------------ | ------------------------------ |
| Increase $B$  | Slope flattens (closer to 0)   | **Counter-clockwise** rotation |
| Decrease $B$  | Slope steepens (more negative) | **Clockwise** rotation         |
| $B = 0$       | Slope undefined                | Vertical line                  |

# Perceptron_Trick

**If a point lies in the positive region of the decision boundary (i.e., the perceptron output is +1), but its true label is -1 (negative class), then:**
- The perceptron has misclassified the point.
- So, to fix this, we subtract a scaled version of that point's feature vector from the weight vector.

**Why subtract?**
 - This shifts the decision boundary away from that point, reducing the likelihood of misclassifying it again.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import  numpy as np
from sklearn.datasets import make_classification
```

```python
X, y = make_classification(n_samples=100,n_features=2,n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,hypercube=False,class_sep=10,random_state=41)
X.shape
```

```python
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='bwr', edgecolor='k')
plt.title("Synthetic Classification Dataset")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.grid(True)
plt.show()
```

```python
def step(z):  ## step function 
    return 1 if z>0 else 0
```

```python
def perceptron(X,y):
    X=np.insert(X,0,1,axis=1)## np.insert(arr, index, values, axis)
    weights=np.ones(X.shape[1])
    lr=0.1

    for i in range(1000):
        j=np.random.randint(0,100)
        y_hat=step(np.dot(X[j],weights))
        weights=weights+lr*(y[j]-y_hat)*X[j]

    return weights[0],weights[1:]
```

```python
intercept_,coef_=perceptron(X,y)
```

```python
print(intercept_,coef)
```

```python
m=-(coef_[0]/coef_[1])
b=-(intercept_/coef_[1])
```

```python
x_input=np.linspace(-3,3,100)
y_input=m*x_input +b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
